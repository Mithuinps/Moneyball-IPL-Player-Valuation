```python
import pandas as pd
import os

auction = pd.read_excel("/content/Auction.xlsx")
batsmen = pd.read_excel("/content/Top_100_batsman.xlsx")
bowlers = pd.read_excel("/content/Top_100_bowlers.xlsx")
matches = pd.read_csv("/content/matches.csv")

print("All four datasets loaded successfully!")
```

```python
datasets = {
      "Auction": auction,
      "Top 100 Batsmen": batsmen,
      "Top 100 Bowlers": bowlers,
      "IPL Matches": matches
 }

for name, data in datasets.items():
    print("\n" + "=" * 50)
    print("DATASET:", name)
    print("Shape (Rows, Columns):", data.shape)
    print("\nColumn Names:")
    print(data.columns.tolist())
    print("\nFirst 5 Rows:")
    display(data.head())


```

```python
for name, data in datasets.items():
    print("\n" + "=" * 50)
    print("DATASET:", name)
    print("Rows and Columns:", data.shape)

    print("\nMissing Values:")
    print(data.isnull().sum())

    print("\nDuplicate Rows:")
    print(data.duplicated().sum())

    print("\nData Types:")
    print(data.dtypes)
```

```python
# Week 1: Summary Statistics

print("AUCTION PRICE SUMMARY")
display(auction.describe())

print("\nTOP BATSMEN PERFORMANCE SUMMARY")
display(batsmen.describe())

print("\nTOP BOWLERS PERFORMANCE SUMMARY")
display(bowlers.describe())

print("\nIPL MATCHES SUMMARY")
display(matches.describe())
```

```python
print("TOP 5 MOST EXPENSIVE PLAYERS IN AUCTION")

print(auction.sort_values(
    by="PRICE(In crore Indian Rupees) ",
        ascending=False
        ).head(5))
print("\nTOP 5 BATSMEN BY RUNS")

print(batsmen.sort_values(
      by="Runs",
          ascending=False
          )[["PLAYER", "Runs"]].head(5))
print("\nTOP 5 BOWLERS BY WICKETS")

print(bowlers.sort_values(
    by="Wkts",
        ascending=False
)[["PLAYER", "Wkts"]].head(5))
```

```python
# Week 1: IPL Team Performance Analysis

print("TOP 10 TEAMS BY MATCH WINS")

team_wins = matches["winner"].value_counts().head(10)

print(team_wins)

print("\nMOST SUCCESSFUL TEAM")

print("Team:", team_wins.index[0])
print("Total Wins:", team_wins.iloc[0])
```
